```python
import anndata as ad
```

```python
import os
spatial_dir = '../data/anndata/'
files = [f for f in sorted(os.listdir(spatial_dir)) if f.endswith('filtered.h5ad')]

adata = ad.concat([ad.read_h5ad(os.path.join(spatial_dir,f)) for f in files])
```

```python
adata.obs
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig,axes = plt.subplots(ncols=2,nrows=2,figsize=(14,14),dpi=150)
samples = adata.obs['sample'].unique()
for i,sample in enumerate(samples):
    sample_adata = adata[adata.obs['sample']==sample].copy()
    ax = axes.flat[i]
    sns.scatterplot(data=sample_adata.obs, x='pxl_row_in_fullres',y='pxl_col_in_fullres',hue='kmeans',ax=ax,palette='deep')
    ax.set_title(sample)

fig,axes = plt.subplots(ncols=2,nrows=2,figsize=(14,14),dpi=150)
samples = adata.obs['sample'].unique()
for i,sample in enumerate(samples):
    sample_adata = adata[adata.obs['sample']==sample].copy()
    ax = axes.flat[i]
    sns.scatterplot(data=sample_adata.obs, x='pxl_row_in_fullres',y='pxl_col_in_fullres',hue='phenotypes',ax=ax,palette='deep')
    ax.set_title(sample)


fig,axes = plt.subplots(ncols=2,nrows=2,figsize=(14,14),dpi=150)
samples = adata.obs['sample'].unique()
for i,sample in enumerate(samples):
    sample_adata = adata[adata.obs['sample']==sample].copy()
    ax = axes.flat[i]
    sns.scatterplot(data=sample_adata.obs, x='pxl_row_in_fullres',y='pxl_col_in_fullres',hue='leiden_clusters',ax=ax,palette='deep')
    ax.set_title(sample)

fig,axes = plt.subplots(ncols=2,nrows=2,figsize=(14,14),dpi=150)
samples = adata.obs['sample'].unique()
for i,sample in enumerate(samples):
    sample_adata = adata[adata.obs['sample']==sample].copy()
    ax = axes.flat[i]
    sns.scatterplot(data=sample_adata.obs, x='pxl_row_in_fullres',y='pxl_col_in_fullres',hue='nCount_Spatial',ax=ax,palette='viridis')
    ax.set_title(sample)
# sns.scatterplot(data=adata.obs, x='pxl_row_in_fullres',y='pxl_col_in_fullres',hue='phenotypes')
```
